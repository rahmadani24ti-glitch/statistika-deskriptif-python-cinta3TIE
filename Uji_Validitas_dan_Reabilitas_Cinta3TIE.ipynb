1. Import Library

```python
# Import library yang dibutuhkan

import pandas as pd
import numpy as np
import re

from scipy.stats import pearsonr, t
from google.colab import files
from IPython.display import display
```

2. Upload file Excel

```python
# Upload file hasil kuesioner
uploaded = files.upload()

# Membaca file Excel
nama_file = list(uploaded.keys())[0]
df = pd.read_excel(nama_file)

# Menampilkan informasi data
print("Data berhasil dibaca!")
print("Jumlah responden:", len(df))
print("Jumlah seluruh kolom:", len(df.columns))

display(df.head())
```

3. Identifikasi kolom pernyataan

```python
# Mengambil kolom pernyataan berdasarkan format nama kolom

kolom_item = [
    kolom for kolom in df.columns
    if re.match(r"^\s*.+?\s*\[.+\]\s*$", str(kolom))
]

# Membuat dataframe khusus butir pernyataan
data_item = df[kolom_item].copy()

print("Jumlah responden:", len(data_item))
print("Jumlah butir pernyataan:", len(kolom_item))

print("\nDaftar butir pernyataan:")

for i, kolom in enumerate(kolom_item, 1):
    print(f"{i}. {kolom}")
```

4. Konversi jawaban skala Likert

```python
# Skala Likert
skala_likert = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5
}

# Mengubah jawaban menjadi angka
data = data_item.replace(skala_likert)

# Konversi menjadi numerik
data = data.apply(pd.to_numeric, errors="coerce")

# Memeriksa data kosong atau gagal dikonversi
print("Jumlah data kosong:", data.isna().sum().sum())

display(data.head())

# Menghentikan proses jika ada data kosong
if data.isna().sum().sum() > 0:
    raise ValueError(
        "Terdapat jawaban kosong atau tidak dikenali. "
        "Periksa kembali data dan label skala Likert."
    )
```

5. Mengelompokkan butir berdasarkan dimensi

```python
# Membuat kelompok dimensi
kelompok_dimensi = {}

for kolom in kolom_item:

    hasil = re.match(r"^\s*(.*?)\s*\[", str(kolom))

    if hasil:
        nama_dimensi = hasil.group(1).strip().rstrip(".").strip()

        if nama_dimensi not in kelompok_dimensi:
            kelompok_dimensi[nama_dimensi] = []

        kelompok_dimensi[nama_dimensi].append(kolom)

# Menampilkan hasil pengelompokan
print("HASIL PENGELOMPOKAN DIMENSI")

for dimensi, item in kelompok_dimensi.items():
    print(f"\n{dimensi}")
    print("Jumlah butir:", len(item))

    for kolom in item:
        print("-", kolom)
```

6. Menghitung nilai r tabel

```python
# Jumlah responden
n = len(data)

# Derajat kebebasan
df_bebas = n - 2

# Taraf signifikansi 5%, dua arah
t_tabel = t.ppf(1 - 0.05/2, df_bebas)

# Menghitung r tabel
r_tabel = t_tabel / np.sqrt(
    df_bebas + t_tabel**2
)

print("Jumlah responden:", n)
print("Derajat kebebasan:", df_bebas)
print("Taraf signifikansi: 0.05")
print("Nilai r tabel:", round(r_tabel, 4))
```

7. Uji Validitas Pearson

```python
# Menampung hasil uji validitas
hasil_validitas = []

for dimensi, item in kelompok_dimensi.items():

    # Menghitung total skor dimensi
    total_skor = data[item].sum(axis=1)

    for i, kolom in enumerate(item, start=1):

        # Menghitung korelasi Pearson
        r_hitung, p_value = pearsonr(
            data[kolom],
            total_skor
        )

        # Corrected item-total correlation
        corrected_r, _ = pearsonr(
            data[kolom],
            total_skor - data[kolom]
        )

        # Menentukan hasil validitas
        if r_hitung > r_tabel:
            keterangan = "Valid"
        else:
            keterangan = "Tidak Valid"

        hasil_validitas.append({
            "Dimensi": dimensi,
            "Butir": f"B{i}",
            "Pernyataan": kolom.split("[", 1)[1].rsplit("]", 1)[0],
            "r hitung": r_hitung,
            "r tabel": r_tabel,
            "Corrected r": corrected_r,
            "p-value": p_value,
            "Keterangan": keterangan
        })

# Membuat tabel hasil validitas
tabel_validitas = pd.DataFrame(hasil_validitas)

display(tabel_validitas.round(4))
```

8. Interpretasi hasil uji validitas

```python
# Menghitung jumlah butir valid dan tidak valid

jumlah_valid = (
    tabel_validitas["Keterangan"] == "Valid"
).sum()

jumlah_tidak_valid = (
    tabel_validitas["Keterangan"] == "Tidak Valid"
).sum()

print("HASIL UJI VALIDITAS")
print("===================")

print("Jumlah seluruh butir:", len(tabel_validitas))
print("Jumlah butir valid:", jumlah_valid)
print("Jumlah butir tidak valid:", jumlah_tidak_valid)

print("\nPersentase valid:",
      round(jumlah_valid / len(tabel_validitas) * 100, 2), "%")

print("Persentase tidak valid:",
      round(jumlah_tidak_valid / len(tabel_validitas) * 100, 2), "%")

print("\nButir yang tidak valid:")

display(
    tabel_validitas[
        tabel_validitas["Keterangan"] == "Tidak Valid"
    ]
)
```

9. Fungsi perhitungan Cronbach's Alpha

```python
# Fungsi menghitung Cronbach's Alpha

def cronbach_alpha(data_item):

    jumlah_item = data_item.shape[1]

    if jumlah_item < 2:
        return np.nan

    varians_item = data_item.var(axis=0, ddof=1).sum()

    varians_total = data_item.sum(axis=1).var(ddof=1)

    if varians_total == 0:
        return np.nan

    alpha = (
        jumlah_item / (jumlah_item - 1)
    ) * (
        1 - varians_item / varians_total
    )

    return alpha
```

10. Uji Reliabilitas setiap dimensi

```python
# Menampung hasil reliabilitas
hasil_reliabilitas = []

for dimensi, item in kelompok_dimensi.items():

    # Mengambil data berdasarkan dimensi
    data_dimensi = data[item]

    # Menghitung Cronbach's Alpha
    alpha = cronbach_alpha(data_dimensi)

    # Menentukan kategori
    if alpha >= 0.90:
        kategori = "Sangat Baik"
    elif alpha >= 0.80:
        kategori = "Baik"
    elif alpha >= 0.70:
        kategori = "Reliabel"
    else:
        kategori = "Tidak Reliabel"

    hasil_reliabilitas.append({
        "Dimensi": dimensi,
        "Jumlah Butir": len(item),
        "Cronbach's Alpha": alpha,
        "Kategori": kategori
    })

# Membuat tabel reliabilitas
tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)

display(tabel_reliabilitas.round(4))
```

11. Menghitung Alpha if Item Deleted

```python
# Menampung hasil alpha jika item dihapus
hasil_alpha_deleted = []

for dimensi, item in kelompok_dimensi.items():

    # Hanya dilakukan jika jumlah item lebih dari 2
    if len(item) > 2:

        for i, kolom in enumerate(item, start=1):

            # Menghapus satu butir
            data_hapus = data[item].drop(columns=[kolom])

            # Menghitung alpha baru
            alpha_baru = cronbach_alpha(data_hapus)

            hasil_alpha_deleted.append({
                "Dimensi": dimensi,
                "Butir": f"B{i}",
                "Alpha jika Dihapus": alpha_baru
            })

tabel_alpha_deleted = pd.DataFrame(hasil_alpha_deleted)

display(tabel_alpha_deleted.round(4))
```

12. Analisis keputusan setiap butir

```python
# Menggabungkan hasil validitas dan alpha jika item dihapus

hasil_analisis = tabel_validitas.merge(
    tabel_alpha_deleted,
    on=["Dimensi", "Butir"],
    how="left"
)

# Mengambil alpha awal setiap dimensi
alpha_awal = dict(zip(
    tabel_reliabilitas["Dimensi"],
    tabel_reliabilitas["Cronbach's Alpha"]
))

def keputusan(row):

    alpha_sekarang = alpha_awal[row["Dimensi"]]
    alpha_hapus = row["Alpha jika Dihapus"]

    if row["Keterangan"] == "Tidak Valid":
        return "Perlu direvisi atau dihapus dan diuji ulang"

    elif pd.notna(alpha_hapus) and alpha_hapus > alpha_sekarang:
        return "Valid, tetapi perlu ditinjau"

    else:
        return "Dipertahankan"

hasil_analisis["Keputusan"] = hasil_analisis.apply(
    keputusan,
    axis=1
)

display(hasil_analisis.round(4))
```

13. Pengujian pengaruh jumlah sampel

```python
# Mengambil 20 responden secara acak

data_20 = data.sample(
    n=20,
    random_state=42
)

# Menghitung r tabel untuk 20 responden
n_20 = len(data_20)
df_20 = n_20 - 2

t_20 = t.ppf(0.975, df_20)

r_tabel_20 = t_20 / np.sqrt(
    df_20 + t_20**2
)

hasil_sampel = []

for dimensi, item in kelompok_dimensi.items():

    total_skor = data_20[item].sum(axis=1)

    for i, kolom in enumerate(item, start=1):

        r_hitung, p_value = pearsonr(
            data_20[kolom],
            total_skor
        )

        hasil_sampel.append({
            "Dimensi": dimensi,
            "Butir": f"B{i}",
            "r hitung": r_hitung,
            "r tabel": r_tabel_20,
            "Keterangan": (
                "Valid" if r_hitung > r_tabel_20
                else "Tidak Valid"
            )
        })

tabel_sampel = pd.DataFrame(hasil_sampel)

print("Jumlah sampel:", n_20)
print("r tabel:", round(r_tabel_20, 4))

display(tabel_sampel.round(4))
```

14. Membandingkan hasil pengujian

```python
# Membandingkan hasil pengujian 38 responden dan 20 responden

perbandingan = tabel_validitas[
    ["Dimensi", "Butir", "r hitung", "Keterangan"]
].merge(
    tabel_sampel[
        ["Dimensi", "Butir", "r hitung", "Keterangan"]
    ],
    on=["Dimensi", "Butir"],
    suffixes=("_38_Responden", "_20_Responden")
)

display(perbandingan.round(4))
```

15. Kesimpulan otomatis pengujian

```python
print("========================================")
print("KESIMPULAN UJI VALIDITAS DAN RELIABILITAS")
print("========================================")

print("\nUJI VALIDITAS")
print("Jumlah responden:", len(data))
print("Jumlah butir:", len(tabel_validitas))
print("r tabel:", round(r_tabel, 4))
print("Butir valid:", jumlah_valid)
print("Butir tidak valid:", jumlah_tidak_valid)

print("\nUJI RELIABILITAS")

for _, baris in tabel_reliabilitas.iterrows():

    print(
        baris["Dimensi"],
        "=> Alpha:",
        round(baris["Cronbach's Alpha"], 4),
        "|",
        baris["Kategori"]
    )

jumlah_reliabel = (
    tabel_reliabilitas["Cronbach's Alpha"] >= 0.70
).sum()

print("\nJumlah dimensi reliabel:",
      jumlah_reliabel,
      "dari",
      len(tabel_reliabilitas))
```

16. Ekspor seluruh hasil ke Excel

```python
# Nama file hasil
nama_output = "Hasil_Lengkap_Uji_Validitas_Reliabilitas.xlsx"

# Menyimpan semua hasil pengujian
with pd.ExcelWriter(nama_output, engine="openpyxl") as writer:

    data.to_excel(
        writer,
        sheet_name="Data Likert",
        index=False
    )

    tabel_validitas.to_excel(
        writer,
        sheet_name="Uji Validitas",
        index=False
    )

    tabel_reliabilitas.to_excel(
        writer,
        sheet_name="Uji Reliabilitas",
        index=False
    )

    tabel_alpha_deleted.to_excel(
        writer,
        sheet_name="Alpha Item Deleted",
        index=False
    )

    hasil_analisis.to_excel(
        writer,
        sheet_name="Analisis Butir",
        index=False
    )

    tabel_sampel.to_excel(
        writer,
        sheet_name="Uji Sampel 20",
        index=False
    )

    perbandingan.to_excel(
        writer,
        sheet_name="Perbandingan Sampel",
        index=False
    )

print("Seluruh pengujian berhasil disimpan!")

# Download file
files.download(nama_output)
```
